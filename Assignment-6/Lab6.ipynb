```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a. 
corrected_temperature = temperature + 2

print("Q1(a) Corrected Temperature:")
print(corrected_temperature)


# b. 
# Formula: F = (9/5)C + 32
fahrenheit = (corrected_temperature * 9 / 5) + 32

print("\nQ1(b) Temperature in Fahrenheit:")
print(fahrenheit)


# c. 
readings_above_32 = corrected_temperature[corrected_temperature > 32]

print("\nQ1(c) Readings greater than 32°C:")
print(readings_above_32)


# d. 
count_above_32 = np.sum(corrected_temperature > 32)

print("\nQ1(d) Number of readings greater than 32°C:")
print(count_above_32)


# e. 
print("Vectorization allows NumPy to perform operations on the")
print("entire array at once without explicitly using a for loop.")
print("Boolean indexing creates a Boolean condition for each")
print("element and selects only the elements satisfying that condition.")
```

Q2

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# (a) Total steps
print("Total steps:", np.sum(steps))

# (b) Mean steps
print("Mean steps:", np.mean(steps))

# (c) Maximum and minimum
print("Maximum:", np.max(steps))
print("Minimum:", np.min(steps))

# (d) Total steps for each day
print("Daily totals:", np.sum(steps, axis=0))

# (e) Total steps for each user
print("User totals:", np.sum(steps, axis=1))

# (f) Position of maximum value
index = np.argmax(steps)
position = np.unravel_index(index, steps.shape)

print("Flattened index:", index)
print("User/day position:", position)
print("Maximum value:", steps[position])
```

Q3

```python
import numpy as np

# (a) Create original array
original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)

# (b) Create slice from index 1 to 4
subset = original[1:5]
print("Subset:", subset)

# (c) Modify the slice
subset[0] = 999
print("Original after modifying subset:", original)
print("Subset:", subset)

# (d) Create an independent copy
original = np.array([1, 2, 3, 4, 5, 6])
copied = original[1:5].copy()
copied[0] = 500

print("Original after modifying copy:", original)
print("Copied array:", copied)

# (e) Create a 3 x 4 matrix
arr = np.arange(1, 13)
matrix = arr.reshape(3, 4)

print("Matrix:\n", matrix)

# (f) Extract required elements
print("First row:", matrix[0, :])
print("Last row:", matrix[-1, :])
print("Second column:", matrix[:, 1])

print("Rows 1-2 and columns 2-3:\n",
      matrix[1:3, 2:4])

# (g) Flatten using both methods
matrix = np.arange(1, 13).reshape(3, 4)

flat1 = matrix.flatten()
flat2 = matrix.ravel()

print("Flatten:", flat1)
print("Ravel:", flat2)

# (h) Modify ravel result
flat2[0] = 100
print("After modifying ravel:")
print("Ravel array:", flat2)
print("Original matrix:\n", matrix)

# (i) Modify flatten result
matrix = np.arange(1, 13).reshape(3, 4)
flat1 = matrix.flatten()

flat1[0] = 500
print("After modifying flatten:")
print("Flatten array:", flat1)
print("Original matrix:\n", matrix)

# (j)
matrix = np.arange(1, 13).reshape(3, 4)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

Q4

```python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# (a) Shape and dimensions
print("Shape:", X.shape)
print("Dimensions:", X.ndim)

# (b) Transpose
XT = X.T
print("Transpose:\n", XT)

# (c) Matrix product
product = XT @ X
print("X.T @ X:\n", product)

# (d) Matrix inverse
inverse = np.linalg.inv(product)
print("Inverse:\n", inverse)

# (e) OLS coefficients
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("OLS coefficients:", beta)

# (f) Interpret coefficients
print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])

# (g) Predict assistance score for a new user
new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("Predicted assistance score:", predicted_score)
```
