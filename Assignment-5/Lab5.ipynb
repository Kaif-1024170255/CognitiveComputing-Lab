```python
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]

print("Original array:", arr)
print("Reversed array:", reversed_arr)


x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)

max_count = np.max(counts)

most_frequent = values[counts == max_count]

print("Array:", x)
print("Most frequent value:", most_frequent)
print("Frequency:", max_count)

for value in most_frequent:
    indices = np.where(x == value)[0]
    print("Indices of", value, ":", indices)
```

```python
import numpy as np

K = np.linspace(10, 100, 25)

print("Array:")
print(K)

print("\nNumber of dimensions:", K.ndim)
print("Shape:", K.shape)
print("Total number of elements:", K.size)
print("Data type:", K.dtype)
print("Total bytes consumed:", K.nbytes)

transposed = K.reshape(1, 25)

print("\nTranspose using reshape():")
print(transposed)

transposed_T = K.T

print("\nUsing T attribute:")
print(transposed_T)
```

```python
import numpy as np
ucs420_Kaif = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array:")
print(ucs420_Kaif)

print("Mean:", np.mean(ucs420_Kaif))
print("Median:", np.median(ucs420_Kaif))
print("Maximum:", np.max(ucs420_Kaif))
print("Minimum:", np.min(ucs420_Kaif))
print("Unique elements:", np.unique(ucs420_Kaif))

reshaped_ucs420_Kaif = ucs420_Kaif.reshape(4, 3)

print("\nReshaped array:")
print(reshaped_ucs420_Kaif)

resized_ucs420_Kaif = np.resize(ucs420_Kaif, (2, 3))

print("\nResized array:")
print(resized_ucs420_Kaif)
```
