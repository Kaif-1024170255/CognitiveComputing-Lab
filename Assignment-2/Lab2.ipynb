```python
import random
print(3," ",1)
L=[]
for i in range(100):
    L.append(random.randrange(100,900))
    
co=0
ce=0
e=[]
o=[]
for i in L:
    if i%2!=0: 
        o.append(i)
        co=co+1
    else:
        e.append(i)
        ce=ce+1
print(3," ",3)
print(ce," ",e)
print(3," ",2)
print(co," ",o)

print(3," ",4)
p=0
primes=[]
def checkPrime(n):
    for i in range(2,n):
        if n%i==0: return False
    return True
for i in L:
    if(checkPrime(i)):
        p=p+1
        primes.append(i)
print(p," ",primes)

print(3," ",5)
```
