```python
my_dict={
    "name":"Kaif Ahmad",
    "roll_no":"1024170255",
    "branch":"COPC",
    "age":"21",
    "city":"Jamshedpur"
}
print(5," ",1)
my_dict["location"] = my_dict.pop("city")
print(my_dict)

print(5," ",2)
my_dict["cgpa"] = 9.2
print(my_dict)

print(5," ",3)
my_dict["age"] += 1
print(my_dict)

print(5," ",4)
dict_pop = my_dict.copy()
removed_branch = dict_pop.pop("branch")
print(dict_pop)
print("Removed value:", removed_branch)

dict_del = my_dict.copy()
del dict_del["branch"]

print("Dictionary after del:")
print(dict_del)

print("Difference:")
print("pop() removes the key and returns its value, whereas del only removes the key and returns nothing.")

print(5," ",5)
print("Key-Value Pairs:")
for key, value in my_dict.items():
    print(f"{key} → {value}")

print(5," ",6)
print("Checking Email:")
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email key does not exist.")

print(5," ",7)
friend_dict = {
    "name": "Banerjee",
    "roll_no": "1024170304",
    "branch": "COE",
    "age": 21,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print("Merged Dictionary:")
print(merged_dict)

print("Explanation:")
print("When duplicate keys exist, the values from the second dictionary (friend_dict) overwrite those from the first.")

print(5," ",8)
string_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}

print("Dictionary with Only String Values:")
print(string_dict)
```

```python
L=[1,0,2,4,1,7,0,2]
A=set([i*7 for i in L])
B=set([i*9 for i in L])
print(4," ",1)
print(A.union(B))
print(4," ",2)
print(A.intersection(B))
print(4," ",4)
print(A.difference(B))
print(B.difference(A))
print(4," ",5)
print(A.symmetric_difference(B))
print(4," ",6)
print(A.issubset(B))
print(B.issuperset(A))
print(4," ",7)
n=int(input("Enter n : "))
try:
    A.remove(n)
    print("Removed:",n)
except KeyError:
    print("Error")
```

```python
import random
print(3," ",1)
L=[]
for i in range(100):
    L.append(random.randrange(100,900))
    
co=0
ce=0
e=[]
o=[]
for i in L:
    if i%2!=0: 
        o.append(i)
        co=co+1
    else:
        e.append(i)
        ce=ce+1
print(3," ",3)
print(ce," ",e)
print(3," ",2)
print(co," ",o)

print(3," ",4)
p=0
primes=[]
def checkPrime(n):
    for i in range(2,n):
        if n%i==0: return False
    return True
for i in L:
    if(checkPrime(i)):
        p=p+1
        primes.append(i)
print(p," ",primes)

print(3," ",5)
print(max(L)," ",L.count(max(L)))
```
